<details>
<summary>👁️ Explication (cliquez après avoir prédit ET exécuté !)</summary>

| Expression | Python | C++ et Java, avec des entiers |
|---|---|---|
| `7 / 2` | `3.5` : `/` donne toujours un nombre à virgule | `3` : la division de deux entiers est entière |
| `7 // 2` | `3` : `//` est la division entière | c'est `7 / 2` qui donne `3` |

En Python, `//` et `%` restent cohérents entre eux : `a == (a // b) * b + a % b`, pour tous les entiers. C'est ce qui fait de `%` le bon outil pour un calcul qui « fait le tour », même avec des négatifs : un jour de la semaine, un indice dans une liste circulaire.

**À retenir :** quand vous traduisez un calcul de C++ ou de Java, remplacez la division entière par `//`, et méfiez-vous des nombres négatifs.

#### 🔬 Mesurons : `in` sur une liste ou sur un set ?

« Test d'appartenance très rapide » : ne le croyez pas sur parole, **mesurez-le**. Le module [`timeit`](https://docs.python.org/fr/3/library/timeit.html) exécute un bout de code un grand nombre de fois et renvoie le temps total.

Avant d'exécuter la cellule, pariez : sur 100 000 éléments, le set est-il 2 fois, 100 fois ou 10 000 fois plus rapide ?

````python
import timeit

# Les mêmes 100 000 entiers, rangés dans une liste et dans un set
NB_ELEMENTS = 100_000
liste = list(range(NB_ELEMENTS))
ensemble = set(liste)
x = NB_ELEMENTS - 1  # Pire cas pour la liste : l'élément cherché est le dernier

NB_RECHERCHES = 200
# lambda: ... crée une petite fonction sans nom, que timeit appelle NB_RECHERCHES fois
temps_liste = timeit.timeit(lambda: x in liste, number=NB_RECHERCHES)
temps_ensemble = timeit.timeit(lambda: x in ensemble, number=NB_RECHERCHES)

print(f"📝 x in liste    : {temps_liste / NB_RECHERCHES * 1e6:10.2f} µs par recherche")
print(
    f"👥 x in ensemble : {temps_ensemble / NB_RECHERCHES * 1e6:10.2f} µs par recherche"
)
print(f"🚀 Le set est environ {temps_liste / temps_ensemble:.0f} fois plus rapide")
````

**💡 Pourquoi un tel écart ?**
- `x in liste` compare `x` à chaque élément, l'un après l'autre : **O(n)**, comme une recherche dans un `std::vector` (C++) ou une `ArrayList` (Java)
- `x in ensemble` calcule le *hash* de `x` et va directement au bon endroit : **O(1)** en moyenne, comme un `std::unordered_set` (C++) ou un `HashSet` (Java)
- Un dictionnaire fonctionne de la même façon pour ses clés

Changez `NB_ELEMENTS` pour 10 000 puis 1 000 000 : le temps de la liste suit, celui du set ne bouge presque pas.

#### 🔮 Prédiction 3 : une grille 3 × 3

````python
grille = [[0] * 3] * 3
grille[0][0] = 1

for rangee in grille:
    print(rangee)
````

<details>
<summary>👁️ Explication (cliquez après avoir prédit ET exécuté !)</summary>

**La règle :** en Python, une variable est un **nom posé sur un objet**, pas une boîte qui contient une valeur. L'affectation `=` ne copie jamais l'objet, elle ajoute un nom.

**Prédiction 1 : `a` et `b` valent tous les deux `[1, 2, 3, 4]`**

`b = a` crée un deuxième nom pour la **même** liste (`a is b` vaut `True`).
- En **Java**, c'est le comportement habituel des objets : `List<Integer> b = a;` ne copie rien non plus
- En **C++**, `std::vector<int> b = a;` ferait une copie. Ce que fait Python correspond plutôt à une référence : `std::vector<int>& b = a;`

Pour une vraie copie : `b = a.copy()` ou `b = list(a)`.

**Prédiction 2 : `['pomme']` puis `['pomme', 'banane']`**

La valeur par défaut `[]` est créée **une seule fois**, au moment où Python exécute le `def`, puis réutilisée à chaque appel.
- En **C++**, l'argument par défaut est réévalué à chaque appel : ce piège n'existe pas. Ici, la liste se comporte comme une variable locale `static`
- En **Java**, il n'y a pas d'argument par défaut : on écrirait une surcharge qui fait `new ArrayList<>()` à chaque appel

L'idiome Python : `None` comme valeur par défaut, et création de la liste dans la fonction (c'est le mini-exercice qui suit).

**Prédiction 3 : les trois rangées valent `[1, 0, 0]`**

`[[0] * 3] * 3` crée **une** liste `[0, 0, 0]`, puis une liste extérieure contenant **trois fois la même référence**.
- En **Java**, c'est comme remplir un `int[][]` avec trois fois le même `int[]`
- En **C++**, c'est comme un `std::vector<std::vector<int>*>` dont les trois pointeurs désignent le même vecteur

La bonne écriture crée une nouvelle rangée à chaque tour : `grille = [[0] * 3 for _ in range(3)]`.

Et pourquoi `[0] * 3` ne pose-t-il pas de problème ? Parce qu'un entier est **immutable** : `grille[0][0] = 1` remplace la référence, il ne modifie pas l'entier `0`.

**À retenir :**

| | Types | Partager une référence |
|---|---|---|
| **Immutables** | `int`, `float`, `str`, `tuple`, `frozenset` | ✅ Sans danger |
| **Mutables** | `list`, `dict`, `set` | ⚠️ Toute modification est visible sous tous les noms |

### 📖 D. Annoter les types

En C++ ou en Java, chaque variable et chaque paramètre a un type déclaré, que le compilateur vérifie. Python est typé **dynamiquement** : une variable n'a pas de type, seule la valeur qu'elle désigne en a un. On peut pourtant **annoter** le type attendu d'un paramètre, ou du résultat d'une fonction :

| Annotation | Se lit |
|---|---|
| `nom: str` | `nom` est une chaîne |

C'est le **typage graduel** : vous annotez ce que vous voulez, et l'outil vérifie ce qui est annoté.

#### 📖 Mode d'emploi des tests

Les tests fonctionnent de la même façon dans tout le notebook.

**Ce qu'est un test.** Chaque méthode `test_...` de la cellule de tests appelle votre fonction avec des valeurs choisies, puis vérifie le résultat avec une **assertion**. Si l'assertion est fausse, le test échoue.

Vous n'avez pas besoin de comprendre toute la syntaxe de la cellule de tests (`class`, `self`, `@unittest.skip`...) : ce sont les lignes `self.assert...` qui disent ce que votre fonction doit faire. Si vous avez déjà fait du Java, c'est le principe de JUnit : `self.assertEqual()` joue le rôle de `assertEquals()`.

**Lire une assertion.** Elle se lit comme une phrase. Dans ce notebook, la valeur obtenue vient toujours en premier, la valeur attendue en second :

| Assertion | Se lit |
|---|---|
| `self.assertEqual(inscrire("Bob", ["Alice"]), ["Alice", "Bob"])` | `inscrire("Bob", ["Alice"])` doit être égal à `["Alice", "Bob"]` |
| `self.assertIs(a, b)` | `a` et `b` doivent être le **même objet** |
| `self.assertIsNot(a, b)` | `a` et `b` doivent être deux objets **distincts** |
| `self.assertTrue(x)` | `x` doit être vrai |
| `self.assertIsNone(x)` | `x` doit valoir `None` |
| `self.assertAlmostEqual(a, b, places=1)` | `a` et `b` doivent être égaux, à 1 chiffre après la virgule près |
| `self.assertCountEqual(a, b)` | `a` et `b` doivent contenir les mêmes éléments, dans n'importe quel ordre |

`assertEqual` compare avec `==`, `assertIs` avec `is`. En Python, `==` compare les valeurs, comme `equals()` en Java, et `is` vérifie qu'il s'agit du même objet, comme `==` entre deux objets Java.

Deux nombres à virgule se comparent avec `assertAlmostEqual`, pas avec `assertEqual` : un float est arrondi en mémoire, si bien que `0.1 + 0.2 == 0.3` vaut `False` (`0.1 + 0.2` donne `0.30000000000000004`).

### 📖 Pour aller plus loin : quand `None` ne suffit pas, les sentinelles et les génériques (bonus)

Dans `inscrire()`, `None` veut dire « aucune liste fournie ». L'astuce marche parce qu'un appelant n'a aucune raison de passer `None` comme liste d'inscrits. Mais quand `None` est une valeur qu'il peut vouloir passer, il faut un autre marqueur : une **sentinelle**, une valeur unique qui ne sert qu'à ça.

Python 3.15 en fabrique avec la fonction native `sentinel()` ([PEP 661](https://peps.python.org/pep-0661/)). Une sentinelle se compare avec `is`, comme `None`, et s'écrit directement dans une annotation de type : `int | None | ABSENT`.

Dans la cellule suivante, `premier()` retourne le premier élément d'une liste, ou `defaut` si la liste est vide. Sans `defaut`, une liste vide est une erreur. Or `defaut` peut valoir `None` : seule une sentinelle distingue « pas de valeur par défaut » de « la valeur par défaut est `None` ».

Le `[T]` qui suit le nom de la fonction la rend **générique**, comme `<T>` en Java : `T` désigne le type des éléments, quel qu'il soit. La même fonction accepte une liste d'entiers ou de chaînes, et l'éditeur sait qu'elle retourne un élément du même type ([PEP 695](https://peps.python.org/pep-0695/), Python 3.12).

#### ⏱️ Durée : 5 min (bonus, si vous êtes en avance)

````python
ABSENT = sentinel("ABSENT")


def premier[T](elements: list[T], defaut: T | None | ABSENT = ABSENT) -> T | None:
    """Premier élément de `elements`, ou `defaut` si la liste est vide."""
    if elements:
        return elements[0]
    if defaut is ABSENT:
        raise ValueError("liste vide, et aucune valeur par défaut")
    return defaut


print(premier([3, 1, 4]))  # 3
print(premier(["kayak", "Laval"]))  # kayak : la même fonction, pour des chaînes
print(premier([], None))  # None : la valeur par défaut fournie est None
print(premier([], 0))  # 0

try:
    premier([])
except ValueError as erreur:
    print(f"ValueError : {erreur}")
````

````python
def comparer_ateliers(
    atelier_a: list[str], atelier_b: list[str]

    """
    Compare les inscrits de deux ateliers.
````

### 🔬 Pour aller plus loin : une file d'attente, liste ou deque ? (bonus)

Une **file d'attente** sert les éléments dans leur ordre d'arrivée : on ajoute à la fin, on retire au début. Avec une liste, retirer le premier élément (`liste.pop(0)`) oblige à décaler tous les autres d'une case : c'est **O(n)**, comme `ArrayList.remove(0)` en Java. Le module `collections` fournit une structure faite pour ça, la **deque** (*double-ended queue*), où retirer au début est en **O(1)**, comme avec `ArrayDeque` en Java.

Avant d'exécuter la cellule, pariez : pour vider une file de 50 000 éléments, la deque est-elle 2 fois, 20 fois ou 200 fois plus rapide ?

#### ⏱️ Durée : 5 min (bonus, si vous êtes en avance)

````python
import timeit
from collections import deque

NB_ELEMENTS = 50_000


def vider_liste() -> None:
    file = list(range(NB_ELEMENTS))
    while file:
        file.pop(0)  # décale tous les éléments restants : O(n)


def vider_deque() -> None:
    file = deque(range(NB_ELEMENTS))
    while file:
        file.popleft()  # O(1)


temps_liste = timeit.timeit(vider_liste, number=1)
temps_deque = timeit.timeit(vider_deque, number=1)
print(f"📝 liste : {temps_liste * 1000:8.1f} ms")
print(f"🚀 deque : {temps_deque * 1000:8.1f} ms")
print(f"La deque est environ {temps_liste / temps_deque:.0f} fois plus rapide")
````

**💡 Pourquoi un tel écart ?** Vider la file demande `n` retraits. Avec une liste, chacun coûte O(n), soit **O(n²)** au total ; avec une deque, chacun coûte O(1), soit O(n) au total. Doublez `NB_ELEMENTS` pour le vérifier : le temps de la deque double, celui de la liste est multiplié par 4.

**À retenir :** une liste est faite pour ajouter et retirer **à la fin**. Pour une file d'attente, prenez une deque.

#### Aplatir une liste de listes : l'unpacking dans une comprehension (Python 3.15)

L'étoile `*` devant une liste la **déballe** (*unpacking*) : ses éléments sont pris un par un, comme s'ils avaient été écrits à la main. `[*a, *b]` met bout à bout les éléments de `a` et de `b`, et `premier, *reste = [1, 2, 3]` range `1` dans `premier` et `[2, 3]` dans `reste`.

Depuis Python 3.15, `*` s'écrit aussi dans une comprehension ([PEP 798](https://peps.python.org/pep-0798/)) : à chaque tour, les éléments déballés sont versés dans la liste en construction. C'est l'écriture la plus simple pour **aplatir** une liste de listes, là où il fallait deux `for` à la suite, dont l'ordre surprend tout le monde.

La même écriture vaut pour une comprehension de set et pour une expression génératrice, cette comprehension sans crochets qui produit ses valeurs une à une sans construire de liste. Avec `**`, elle fusionne des dictionnaires : `{**d for d in dicos}`. Attention, `**` n'existe que dans une comprehension de dictionnaire : `(**d for d in dicos)` est une erreur de syntaxe.

**⚠️ VS Code souligne ces écritures en rouge, à tort.** Son analyseur de code, Pylance, ne connaît pas encore cette nouveauté de Python 3.15 et les prend pour des erreurs de syntaxe. La cellule suivante s'exécute pourtant sans erreur : les outils suivent le langage avec un peu de retard.

````python
grille = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# Avant Python 3.15 : deux `for` à la suite, à lire de gauche à droite
plat = [case for rangee in grille for case in rangee]
print(f"Deux for  : {plat}")

# ✅ Depuis Python 3.15 : `*rangee` verse les cases de chaque rangée dans la liste
plat = [*rangee for rangee in grille]
print(f"Unpacking : {plat}")

# La même écriture dans une expression génératrice : la somme de toutes les cases,
# sans construire la liste aplatie
print(f"Somme     : {sum(*rangee for rangee in grille)}")

# Et avec `**`, pour fusionner des dictionnaires
groupes = [{"Alice": 12, "Bob": 15}, {"Charlie": 9}]
fusion = {**notes for notes in groupes}
print(f"Fusion    : {fusion}")
````

### 📖 Pour aller plus loin : les générateurs, des valeurs calculées à la demande (bonus)

Une list comprehension calcule toutes ses valeurs d'un coup et les range en mémoire. Avec des parenthèses à la place des crochets, on obtient une **expression génératrice** : elle ne calcule chaque valeur qu'au moment où on la lui demande. C'est l'**évaluation paresseuse**, celle des `Stream` de Java. Vous en avez déjà croisé une plus haut : `sum(*rangee for rangee in grille)`.

Une fonction qui contient `yield` produit elle aussi ses valeurs une à une : c'est une **fonction génératrice**.

#### ⏱️ Durée : 5 min (bonus, si vous êtes en avance)

````python
import sys
from collections.abc import Iterator

carres_liste = [x * x for x in range(1_000_000)]
carres_generateur = (x * x for x in range(1_000_000))

# sys.getsizeof() mesure l'objet lui-même : pour la liste, son million de
# références (8 octets chacune), sans compter les entiers qu'elles désignent
print(f"Liste      : {sys.getsizeof(carres_liste):>10_} octets")
print(f"Générateur : {sys.getsizeof(carres_generateur):>10_} octets")

# Un générateur ne se parcourt qu'une fois : la seconde somme ne trouve plus rien
print(f"Première somme : {sum(carres_generateur)}")
print(f"Seconde somme  : {sum(carres_generateur)}")


def compte_a_rebours(depart: int) -> Iterator[int]:
    """Produit depart, depart - 1, ..., 1, une valeur à la fois."""
    while depart > 0:
        yield depart
        depart -= 1


print(list(compte_a_rebours(5)))
````

**À retenir :** quand on ne parcourt les valeurs qu'une fois, par exemple pour les additionner, une expression génératrice évite de construire une liste en mémoire. `range()` est lui-même paresseux : `range(10**12)` ne coûte rien tant qu'on ne le parcourt pas. L'annotation `Iterator[int]` dit que la fonction produit des entiers, un par un.

### 📖 D. Les fonctions sont des valeurs : `lambda` et fonctions d'ordre supérieur

En Python, une fonction est une valeur comme une autre : on peut la ranger dans une variable, la passer en argument à une autre fonction, ou la retourner. Une fonction qui reçoit ou retourne une fonction est dite **d'ordre supérieur**. Vous en avez déjà utilisé une sans le savoir : en partie 1, `timeit.timeit()` reçoit la fonction à chronométrer. Java fait de même depuis Java 8, avec ses lambdas et ses références de méthodes :

| Python | Java |
|---|---|
| `sorted(mots, key=len)` | `mots.sort(Comparator.comparing(String::length))` |
| `lambda paire: paire[1]` | `paire -> paire.getValue()` |
| `[n * 2 for n in notes if n >= 10]` | `notes.stream().filter(n -> n >= 10).map(n -> n * 2).toList()` |

#### ⏱️ Durée : 10 min (démonstration et mini-exercice)

````python
from collections.abc import Callable
from functools import reduce

mots = ["Laval", "kayak", "ressasser", "été"]
resultats = {"Alice": 12, "Bob": 18, "Charlie": 15}

# 1. Une fonction est une valeur : on passe len, sans parenthèses, à sorted()
print(sorted(mots, key=len))  # trie selon la longueur de chaque mot

# 2. max() et min() acceptent aussi une clé : le mot le plus long
print(max(mots, key=len))

# 3. lambda : une petite fonction sans nom, d'une seule expression, écrite sur place
print(sorted(resultats.items(), key=lambda paire: paire[1]))  # trie selon la note

# 4. map() et filter() existent, comme dans les Streams de Java...
print(list(map(str.upper, mots)))  # lisible : la fonction existe déjà
print(list(filter(lambda note: note >= 13, resultats.values())))
# ... mais la comprehension dit la même chose plus clairement
print([note for note in resultats.values() if note >= 13])
print(reduce(lambda a, b: a + b, resultats.values()))  # sum() fait mieux

# Retirez le # de la ligne suivante : ruff la souligne (C417), et propose la
# comprehension équivalente, plus lisible.
# doubles = list(map(lambda note: note * 2, resultats.values()))


# 5. Une fonction qui retourne une fonction
def au_moins(seuil: int) -> Callable[[int], bool]:
    """Retourne une fonction qui teste si une note atteint `seuil`."""
    return lambda note: note >= seuil


est_admis = au_moins(10)
est_mention_bien = au_moins(14)
print(est_admis(12), est_mention_bien(12))
````

**💡 À retenir :**
- Une fonction se passe **sans parenthèses** : `key=len`, et non `key=len()`, qui appellerait la fonction.
- `lambda` convient à une petite fonction passée en argument. Pour lui donner un nom, écrivez plutôt un `def` : ruff signale `carre = lambda x: x * x` (règle `E731`).
- Pour transformer ou filtrer une liste, la comprehension est la forme pythonique. `map()` reste lisible avec une fonction qui existe déjà, comme `map(str.upper, mots)`. `reduce()` a été rangé dans le module `functools` : `sum()`, `max()`, `any()` ou `all()` couvrent la plupart des cas.
- Une fonction qui retourne une fonction garde en mémoire les variables de son contexte, ici `seuil` : c'est une **fermeture** (*closure*). Les décorateurs, comme `@unittest.skip("à activer")`, reposent sur ce mécanisme.
- `Callable[[int], bool]` est le type d'une fonction qui reçoit un `int` et retourne un `bool`, comme `Predicate<Integer>` en Java.

```python
def classer_admis(resultats: dict[str, int], seuil: int = 10) -> list[str]:
    """
    Retourne les noms des admis, de la meilleure note à la moins bonne.

    Args:
        resultats: Dictionnaire {nom: note}
        seuil: Note minimale pour être admis

    Returns:
        list: Les noms des admis ; à note égale, par ordre alphabétique
    """
    admis = [nom for nom, note in resultats.items() if note >= seuil]
    return sorted(admis, key=lambda nom: (-resultats[nom], nom))
```

**Points clés :**
- ✅ La comprehension filtre : elle ne garde que les noms dont la note atteint le seuil
- ✅ `key=` reçoit une fonction, ici une `lambda`, appelée sur chaque nom pour obtenir sa clé de tri
- ✅ La clé est un tuple `(-note, nom)` : Python trie d'abord sur la note, puis sur le nom en cas d'égalité
- ✅ `-note` inverse l'ordre des notes sans inverser celui des noms. Avec `reverse=True`, les noms seraient eux aussi rangés à l'envers

```python
import unicodedata

def est_palindrome(texte: str) -> bool:
    """
    Vérifie si un texte est un palindrome.
    
    Args:
        texte: Chaîne à vérifier
        
    Returns:
        bool: True si palindrome, False sinon
    """
    # Normaliser : enlever accents, passer en minuscules, garder seulement les lettres/chiffres
    normalise = unicodedata.normalize('NFD', texte)
    normalise = ''.join(c for c in normalise if unicodedata.category(c) != 'Mn')
    normalise = ''.join(c.lower() for c in normalise if c.isalnum())
    
    # Comparer avec l'inverse
    return normalise == normalise[::-1]
```

```python
import unicodedata

def est_palindrome(texte: str) -> bool:
    """Version plus concise"""

    normalise = unicodedata.normalize('NFD', texte)
    normalise = re.sub(r'[^a-z0-9]', '', normalise.lower())
    
    return normalise == normalise[::-1]
```

### ✏️ Exercice : Refactoriser l'analyseur de notes

#### 🎯 Objectif

#### 📝 Contexte
Un étudiant a écrit la fonction `analyser_resultats()`, qui lit un fichier de notes d'examen au format CSV. Un fichier CSV (*Comma-Separated Values*) est un fichier texte qui contient une ligne par étudiant, avec des champs séparés par des virgules, par exemple `Alice,15.5`. Elle retourne la moyenne, le nombre d'admis (note ≥ 10) et le nombre de notes valides. #### 📖 Refactoriser, c'est quoi ?
Refactoriser, c'est améliorer l'écriture d'un code **sans changer ce qu'il fait**. Le danger, c'est de casser quelque chose sans s'en apercevoir. La méthode tient en trois règles :
1. **Avant de toucher au code**, des tests vérifient ce qu'il fait aujourd'hui. Ils sont verts : c'est votre **filet de sécurité**
2. **Une seule transformation à la fois**, la plus petite possible
3. **Après chaque transformation**, relancez les tests : ils doivent **rester verts**. S'ils deviennent rouges, vous avez changé le comportement : annulez la dernière modification (Ctrl+Z) et recommencez plus prudemment

C'est la dernière étape du cycle de la partie 3, « refactorez quand tout est vert », appliquée à un programme existant.

````python
# ─────────────────────────────────────────────────────────
# 💻 VOTRE CODE ICI 👇
# ─────────────────────────────────────────────────────────


def analyser_resultats(fichier):
    """
    Analyse les résultats d'examen depuis un fichier CSV.

    Args:
        fichier: Chemin du fichier CSV

    Returns:
        tuple: (moyenne, nb_admis, nb_total)
    """
    f = open(fichier)  # noqa: SIM115
    lignes = f.readlines()
    f.close()

    notes = []
    i = 0
    while i < len(lignes):
        ligne = lignes[i]
        if ligne[0] != "#":
            parties = ligne.split(",")
            if len(parties) >= 2:
                try:
                    note_num = float(parties[1])
                    if note_num >= 0 and note_num <= 20:
                        notes.append(note_num)
                except:  # noqa: E722
                    pass
        i = i + 1

    somme = 0
    for note in notes:
        somme = somme + note
    moyenne = somme / len(notes)

    admis = 0
    for note in notes:
        if note >= 10:
            admis = admis + 1

    return moyenne, admis, len(notes)


# ─────────────────────────────────────────────────────────
````

```python
LETTRES_PAR_VALEUR = {
    1: "AEIOULNRST",
    2: "DG",
    3: "BCMP",
    4: "FHVWY",
    5: "K",
    8: "JX",
    10: "QZ",
}

VALEURS: dict[str, int] = {}
for valeur, lettres in LETTRES_PAR_VALEUR.items():
    for lettre in lettres:
        VALEURS[lettre] = valeur


def score(word: str) -> int:
    return sum(VALEURS[lettre] for lettre in word.upper())
```

**Points clés :**
- ✅ Le dictionnaire `VALEURS` (`{"A": 1, "E": 1, ..., "Z": 10}`) est construit **une seule fois**, en dehors de la fonction, et non à chaque appel
- ✅ `word.upper()` met le mot en majuscules, comme les clés du dictionnaire
- ✅ `sum()` avec une comprehension sans crochets fait la somme des valeurs. Pour un mot vide, il n'y a rien à additionner : `sum()` retourne 0

```python
from collections import Counter


def find_anagrams(word: str, candidates: list[str]) -> list[str]:
    mot = word.lower()
    lettres = Counter(mot)
    return [
        candidat
        for candidat in candidates
        if candidat.lower() != mot and Counter(candidat.lower()) == lettres
    ]
```

**Points clés :**
- ✅ Une comprehension avec un `if` garde les candidats qui conviennent, tels qu'ils sont écrits dans la liste
- ✅ `Counter(mot)` compte chaque lettre, comme à la question 3 des inscriptions. Deux `Counter` sont égaux s'ils ont les mêmes lettres avec les mêmes nombres : c'est exactement la définition d'un anagramme
- ✅ `.lower()` des deux côtés : la casse ne compte ni pour comparer les lettres, ni pour reconnaître le mot lui-même
- ✅ Les lettres du mot sont comptées **une seule fois**, avant la comprehension, et non pour chaque candidat

**Version sans Counter :** deux anagrammes donnent la même liste une fois leurs lettres triées.
```python
def find_anagrams(word: str, candidates: list[str]) -> list[str]:
    mot = word.lower()
    lettres = sorted(mot)
    return [
        candidat
        for candidat in candidates
        if candidat.lower() != mot and sorted(candidat.lower()) == lettres
```
