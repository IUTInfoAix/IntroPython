#### 🔬 Mesurons : `in` sur une liste ou sur un set ?

« Test d'appartenance très rapide » : ne le croyez pas sur parole, **mesurez-le**. Le module [`timeit`](https://docs.python.org/fr/3/library/timeit.html) exécute un bout de code un grand nombre de fois et renvoie le temps total.

Avant d'exécuter la cellule, pariez : sur 100 000 éléments, le set est-il 2 fois, 100 fois ou 10 000 fois plus rapide ?

````python
import timeit

# Les mêmes 100 000 entiers, rangés dans une liste et dans un set
NB_ELEMENTS = 100_000
liste = list(range(NB_ELEMENTS))
ensemble = set(liste)
x = NB_ELEMENTS - 1  # Pire cas pour la liste : l'élément cherché est le dernier

NB_RECHERCHES = 200
temps_liste = timeit.timeit(lambda: x in liste, number=NB_RECHERCHES)
temps_ensemble = timeit.timeit(lambda: x in ensemble, number=NB_RECHERCHES)

print(f"📝 x in liste    : {temps_liste / NB_RECHERCHES * 1e6:10.2f} µs par recherche")
print(
    f"👥 x in ensemble : {temps_ensemble / NB_RECHERCHES * 1e6:10.2f} µs par recherche"
)
print(f"🚀 Le set est environ {temps_liste / temps_ensemble:.0f} fois plus rapide")
````

**💡 Pourquoi un tel écart ?**
- `x in liste` compare `x` à chaque élément, l'un après l'autre : **O(n)**, comme une recherche dans un `std::vector` (C++) ou une `ArrayList` (Java)
- `x in ensemble` calcule le *hash* de `x` et va directement au bon endroit : **O(1)** en moyenne, comme un `std::unordered_set` (C++) ou un `HashSet` (Java)
- Un dictionnaire fonctionne de la même façon pour ses clés

Changez `NB_ELEMENTS` pour 10 000 puis 1 000 000 : le temps de la liste suit, celui du set ne bouge presque pas.

<summary>👁️ Explication (cliquez après avoir prédit ET exécuté !)</summary>

**La règle :** en Python, une variable est un **nom posé sur un objet**, pas une boîte qui contient une valeur. L'affectation `=` ne copie jamais l'objet, elle ajoute un nom.

**Prédiction 1 : `a` et `b` valent tous les deux `[1, 2, 3, 4]`**

`b = a` crée un deuxième nom pour la **même** liste (`a is b` vaut `True`).
- En **Java**, c'est le comportement habituel des objets : `List<Integer> b = a;` ne copie rien non plus
- En **C++**, `std::vector<int> b = a;` ferait une copie. L'équivalent Python serait plutôt `std::vector<int>& b = a;`

Pour une vraie copie : `b = a.copy()` ou `b = list(a)`.

**Prédiction 2 : `['pomme']` puis `['pomme', 'banane']`**

La valeur par défaut `[]` est créée **une seule fois**, au moment où Python exécute le `def`, puis réutilisée à chaque appel.
- En **C++**, l'argument par défaut est réévalué à chaque appel : ce piège n'existe pas. Ici, la liste se comporte comme une variable locale `static`
- En **Java**, il n'y a pas d'argument par défaut : on écrirait une surcharge qui fait `new ArrayList<>()` à chaque appel

L'idiome Python : `None` comme valeur par défaut, et création de la liste dans la fonction (c'est le mini-exercice qui suit).

**Prédiction 3 : les trois rangées valent `[1, 0, 0]`**

`[[0] * 3] * 3` crée **une** liste `[0, 0, 0]`, puis une liste extérieure contenant **trois fois la même référence**.
- En **Java**, c'est comme remplir un `int[][]` avec trois fois le même `int[]`
- En **C++**, c'est comme un `std::vector<std::vector<int>*>` dont les trois pointeurs désignent le même vecteur

La bonne écriture crée une nouvelle rangée à chaque tour : `grille = [[0] * 3 for _ in range(3)]`.

Et pourquoi `[0] * 3` ne pose-t-il pas de problème ? Parce qu'un entier est **immutable** : `grille[0][0] = 1` remplace la référence, il ne modifie pas l'entier `0`.

**À retenir :**

| | Types | Partager une référence |
|---|---|---|
| **Immutables** | `int`, `float`, `str`, `tuple`, `frozenset` | ✅ Sans danger |
| **Mutables** | `list`, `dict`, `set` | ⚠️ Toute modification est visible sous tous les noms |

```python
def inscrire(nom, inscrits=None):
    """
    Ajoute un nom à une liste d'inscrits.

    Args:
        nom: Nom à ajouter
        inscrits: Liste à compléter (une nouvelle liste est créée si absente)

    Returns:
        list: La liste des inscrits, avec le nouveau nom à la fin
    """
    if inscrits is None:
        inscrits = []
    inscrits.append(nom)
    return inscrits
```

**Points clés :**
- ✅ `None` est immutable : le partager entre tous les appels est sans danger
- ✅ `inscrits = []` est exécuté **à chaque appel** : chaque appel sans liste reçoit la sienne
- ✅ `is None` plutôt que `== None` pour tester l'absence de valeur

**⚠️ Le raccourci qui ne marche pas :** `inscrits = inscrits or []` semble équivalent, mais une liste vide compte comme fausse. La liste vide fournie par l'appelant serait remplacée par une autre : c'est ce que détecte le test 5.

### 📖 A. List Comprehensions : écrire des boucles en une ligne

Au lieu d'écrire :
```python
resultat = []
for x in liste:
    resultat.append(x * 2)
```

On peut écrire :
```python
resultat = [x * 2 for x in liste]
```

C'est plus court ET plus lisible !

````python
# ❌ Style classique
carres = []
for x in range(10):
    carres.append(x**2)
print(f"Version boucle : {carres}")

# ✅ Style Python
carres = [x**2 for x in range(10)]
print(f"Version comprehension : {carres}")
````

### 📖 B. Gestion d'erreurs pythonique : EAFP

**EAFP** = "Easier to Ask Forgiveness than Permission"

En Python, on préfère **essayer puis gérer l'erreur** plutôt que **vérifier avant**.

````python
donnees = {"nom": "Alice", "age": 25}

# ❌ Style "Look Before You Leap" (Java/C)
if "email" in donnees:
    email = donnees["email"]
else:
    email = "inconnu@example.com"
print(f"Style LBYL : {email}")

# ✅ Style Python : EAFP
try:
    email = donnees["email"]
except KeyError:
    email = "inconnu@example.com"
print(f"Style EAFP : {email}")

# 🎯 Encore mieux : dict.get()
email = donnees.get("email", "inconnu@example.com")
print(f"Style pythonique : {email}")
````

```python
import unicodedata

def est_palindrome(texte):
    """
    Vérifie si un texte est un palindrome.
    
    Args:
        texte: Chaîne à vérifier
        
    Returns:
        bool: True si palindrome, False sinon
    """
    # Normaliser : enlever accents, passer en minuscules, garder seulement les lettres/chiffres
    normalise = unicodedata.normalize('NFD', texte)
    normalise = ''.join(c for c in normalise if unicodedata.category(c) != 'Mn')
    normalise = ''.join(c.lower() for c in normalise if c.isalnum())
    
    # Comparer avec l'inverse
    return normalise == normalise[::-1]
```

```python
def compresser_rle(texte):
    """
    Compresse un texte avec l'algorithme RLE.
    
    Args:
        texte: Chaîne à compresser
        
    Returns:
        str: Chaîne compressée (format: caractère + nombre de répétitions)
    """
    if not texte:
        return ""
    
    resultat = []
    i = 0
    
    while i < len(texte):
        caractere = texte[i]
        compte = 1
        
        # Compter les répétitions consécutives
        while i + compte < len(texte) and texte[i + compte] == caractere:
            compte += 1
        
        resultat.append(f"{caractere}{compte}")
        i += compte
    
    return ''.join(resultat)
```

```python
import string

def evaluer_mot_de_passe(mdp):
    """
    Évalue la robustesse d'un mot de passe.
    
    Args:
        mdp: Mot de passe à évaluer
        
    Returns:
        int: Score de 0 à 100
    """
    # Trop court : score nul, aucun bonus ne s'applique
    longueur = len(mdp)
    if longueur < 8:
        return 0

    # Score de base selon la longueur
    if longueur < 12:
        score = 20
    elif longueur < 16:
        score = 40
    else:
        score = 60
    
    # Bonus pour la diversité des caractères
    if any(c.islower() for c in mdp):
        score += 10
    if any(c.isupper() for c in mdp):
        score += 10
    if any(c.isdigit() for c in mdp):
        score += 10
    if any(c in string.punctuation for c in mdp):
        score += 10
    
    # Pénalité pour mots communs
    mots_communs = ["password", "123456", "qwerty", "admin"]
    if mdp.lower() in mots_communs:
        score -= 50
    
    # Limiter le score entre 0 et 100
    return max(0, min(100, score))
```

**Version plus concise :**
```python
import string

def evaluer_mot_de_passe(mdp):
    """Version concise"""
    # Trop court : score nul, sans bonus
    longueur = len(mdp)
    if longueur < 8:
        return 0

    # Score de base : 20, 40, puis 60 au maximum
    score = min((longueur - 8) // 4 * 20 + 20, 60)
    
    # Bonus
    checks = [
        any(c.islower() for c in mdp),
        any(c.isupper() for c in mdp),
        any(c.isdigit() for c in mdp),
        any(c in string.punctuation for c in mdp)
    ]
    score += sum(checks) * 10
    
    # Pénalité
    if mdp.lower() in ["password", "123456", "qwerty", "admin"]:
        score -= 50
    
    return max(0, min(100, score))
```

**Points clés :**
- ✅ Moins de 8 caractères : retour immédiat de 0, sans bonus
- ✅ Score progressif selon la longueur
- ✅ `any()` pour vérifier la présence d'au moins un caractère d'un type
- ✅ `string.punctuation` contient tous les caractères spéciaux
- ✅ `max(0, min(100, score))` pour borner le score entre 0 et 100
- ✅ Liste de mots communs à éviter
